Problem Statement

The aim of this project is to develop a Python program that analyzes student marks using NumPy. The program accepts marks for multiple students, stores them in a NumPy array, and calculates total marks, average marks, highest marks, and lowest marks. Based on the calculated average, the program assigns grades using conditional statements.

Logic Explanation

The program begins by asking the user to enter the number of students. It then uses a loop to collect the marks for each student and stores them in a list. After collecting all the marks, the list is converted into a NumPy array so that mathematical operations can be performed efficiently. A user-defined function named calculate_grade() is created to assign grades based on the average marks. The program uses NumPy functions such as np.sum() to calculate the total marks, np.mean() to calculate the average marks, np.max() to find the highest marks, and np.min() to determine the lowest marks. Inside the calculate_grade() function, a conditional block is used to assign grades according to the defined criteria: A for above 85, B for marks between 70 and 85, C for marks between 50 and 69, and Fail for marks below 50. Finally, all the calculated results including total, average, highest, lowest marks, and the assigned grade are displayed clearly to the user.

Complete Code

```python

import numpy as np

def calculate_grade(average_marks):
    if average_marks > 85:
        return "A"
    elif 70 <= average_marks <= 85:
        return "B"
    elif 50 <= average_marks < 70:
        return "C"
    else:
        return "Fail"

number_of_students = int(input("Enter number of students: "))

student_marks_list = []

for student_index in range(number_of_students):
    marks = float(input(f"Enter marks for student {student_index + 1}: "))
    student_marks_list.append(marks)

student_marks_array = np.array(student_marks_list)

total_marks = np.sum(student_marks_array)
average_marks = np.mean(student_marks_array)
highest_marks = np.max(student_marks_array)
lowest_marks = np.min(student_marks_array)

assigned_grade = calculate_grade(average_marks)

print("\n------ Student Marks Analysis ------")
print("Marks Entered:", student_marks_array)
print("Total Marks:", total_marks)
print("Average Marks:", average_marks)
print("Highest Marks:", highest_marks)
print("Lowest Marks:", lowest_marks)
print("Assigned Grade:", assigned_grade)
```
