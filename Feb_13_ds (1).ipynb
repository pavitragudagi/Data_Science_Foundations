
# 2-d and 3-d array



```python
# print a 2-d array with values like[90,80,70,60] [300,200,100,400].
import numpy as np
a2d = np.array([[90,80,70,60],[300,200,100,400]])
print(a2d)
```

```python
print(type(a2d))
```

```python
print(a2d.ndim)
```

```python
print(a2d.shape)
```

```python
# how to create a 3-d array


arr3 = np.array([
    [[1,2,3],[4,5,6]],
    [[10,20,30],[40,50,60]],
    [[0,0,0],[1,1,1]]
    ])

print(arr3)
```

```python
print(type(arr3))
```

```python
print(arr3.ndim)
```

```python
print(arr3.shape)
```

```python
# how to create array using aranger()


arr5 = np.arange(1,10)
print(arr5)
```

```python
arr6 = np.arange(11,50,3)
print(arr6)
```

```python
# how to create a 2 dimensional array using arange and reshape


arr7 = np.arange(1,16).reshape(5,3)
print(arr7)
```

```python
# print an array values between 1 and 100and reshape it as per your choice


import numpy as np

# Create array from 1 to 100
arr = np.arange(1, 101)

# Reshape the array (10 rows, 10 columns)
reshaped_arr = arr.reshape(10, 10)

# Print the reshaped array
print(reshaped_arr)
```

```python
 # print an  array from 101 to 500 and skip 2 digits and reshape it into 5 decsk atleast.
# reshape(d,r,c)


arr9 = np.arange(101,501,2).reshape(5,5,8)
print(arr9)
print(arr9.shape)

```
