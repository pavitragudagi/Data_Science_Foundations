**Q2. reverse a string**

```python
def reverse_string(text):
    reversed_text = ""
    for char in text:
        reversed_text = char + reversed_text
    return reversed_text

print(reverse_string("Python"))
```

**Q3. palindrome number**

```python
def is_palindrome(num):
    original = str(num)
    reversed_num = original[::-1]

    if original == reversed_num:
        return "Palindrome Number"
    else:
        return "Not a Palindrome Number"

print(is_palindrome(121))
print(is_palindrome(123))
```

**Q9. visualization**

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.set(style="whitegrid")

plt.figure(figsize=(8,5))

plt.hist(df["retail_price"].dropna(), bins=30)

plt.title("Distribution of Product Prices")
plt.xlabel("Price")
plt.ylabel("Number of Products")

plt.show()
```

2. **count plot of product categories**

```python
plt.figure(figsize=(10,6))

sns.countplot(y=df["product_category_tree"], order=df["product_category_tree"].value_counts().index)

plt.title("Count of Products in Each Category")
plt.xlabel("Count")
plt.ylabel("Category")

plt.show()
```

**3. box plot-price vs category**

```python
plt.figure(figsize=(10,6))

sns.boxplot(x="retail_price", y="product_category_tree", data=df)

plt.title("Price Distribution Across Categories")

plt.show()
```
