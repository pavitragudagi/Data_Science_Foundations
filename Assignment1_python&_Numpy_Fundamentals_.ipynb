**10**. Slicing is a technique in Python used to extract a portion (subset) of elements from a sequence like a list, string, or array.

**11**. Slicing Works in Python Lists.

Slicing in lists returns a new list containing selected elements.

example:

**12**. Slicing Works in NumPy Arrays

**13**. The range() function in Python is used to generate a sequence of numbers.
It is commonly used in loops, especially in for loops, to control how many times the loop runs.

**Parameters of range()**

1. start:

The number from which the sequence begins.

Default value is 0 if not given.

example:for i in range(5)

    print(i)

output: 0 1 2 3 4

2. stop:

The number at which the sequence stops.

This value is not included in the output.

example:for i in range(2, 6):

    print(i)

output:2 3 4 5

3. step:

The difference between each number in the sequence.

Default value is 1.

example:for i in range(1, 10, 2):

    print(i)
output:1 3 5 7 9




**14**. A palindrome is a word or number that reads the same forward and backward.

**Basic Idea Behind Palindrome Checking**



1.Reverse the given number or string.

2️. Compare the reversed value with the original value.

3️. If both are equal → It is a palindrome.

4️. If not equal → It is not a palindrome.

example:text = "madam"

if text == text[::-1]:

    print("Palindrome")
else:

    print("Not Palindrome")

```python
# Program to check number type

num = int(input("Enter a number: "))

if num > 0:
    print("The number is Positive.")
elif num < 0:
    print("The number is Negative.")
else:
    print("The number is Zero.")
```

```python
def calculate_sum_average(numbers):
    total = 0

    for num in numbers:
        total += num

    average = total / len(numbers)
    return total, average


nums = [10, 20, 30, 40, 50]
total, avg = calculate_sum_average(nums)

print("Sum:", total)
print("Average:", avg)
```

```python
n = int(input("Enter value of n: "))

count = 0
num = 2

while count < n:
    print(num)
    num += 2
    count += 1
```

```python
string = input("Enter a string: ")

reversed_string = ""

for char in string:
    reversed_string = char + reversed_string

if string == reversed_string:
    print("It is a Palindrome.")
else:
    print("It is not a Palindrome.")
```
