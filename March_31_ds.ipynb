Natural Language Processing (NLP) Fundamentals

PART 2: TEXT PREPROCESSING

```python
import re

text = "I loved the new phone!!! It’s absolutely amazing. I'm Going to bye that phone."

# Convert to lowercase
text = text.lower()

# Remove punctuation
text = re.sub(r'[^\w\s]', '', text)

print(text)
```

PART 3: TOKENIZATION

```python
import nltk
nltk.download('punkt')
nltk.download('punkt_tab')
from nltk.tokenize import word_tokenize
 # text = "I love machine learning"
tokens = word_tokenize(text)
print(tokens)
```

PART 4: STOPWORD REMOVAL

```python
from nltk.corpus import stopwords
nltk.download('stopwords')
stop_words = set(stopwords.words('english'))

filtered_tokens = [word for word in tokens if word.lower() not in stop_words]
print(filtered_tokens)
```

PART 5: STEMMING

```python
from nltk.stem import PorterStemmer
stemmer = PorterStemmer()
words = ["playing", "studies", "better"]
stems = [stemmer.stem(word) for word in words]
print(stems)
```

PART 6: LEMMATIZATION

```python
from nltk.stem import WordNetLemmatizer
nltk.download('wordnet')
nltk.download('omw-1.4')
lemmatizer = WordNetLemmatizer()
words = ["playing", "studies", "better"]
lemmas = [lemmatizer.lemmatize(word) for word in words]
print(lemmas)
```

PART 7: BAG OF WORDS (BoW)

```python
from sklearn.feature_extraction.text import CountVectorizer
documents = [
"I love NLP",
"I love Python"
]
vectorizer = CountVectorizer()
X = vectorizer.fit_transform(documents)
print("Vocabulary:", vectorizer.get_feature_names_out())
print("BoW Matrix:\n", X.toarray())
```

PART 8: TF-IDF

```python
from sklearn.feature_extraction.text import TfidfVectorizer
documents = [
"I love NLP",
"I love Python",
"NLP is amazing"
]
tfidf = TfidfVectorizer()
X = tfidf.fit_transform(documents)
print("Vocabulary:", tfidf.get_feature_names_out())
print("TF-IDF Matrix:\n", X.toarray())
```

PART 9: END-TO-END MINI PIPELINE

```python
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split
texts = [
"I love this phone",
"This phone is terrible",
"Amazing battery life",
"Worst purchase ever"
]
labels = [1, 0, 1, 0]
vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(texts)
X_train, X_test, y_train, y_test = train_test_split(X, labels, test_size=0.2,
random_state=42)
model = LogisticRegression()
model.fit(X_train, y_train)
print("Accuracy:", model.score(X_test, y_test))
```
