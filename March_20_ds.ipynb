Z-Test

```python
import numpy as np
from scipy import stats
# sample data
data = [52,55,58,60,62,65,66,68,70,72]
# population mean
population_mean = 60
# perform one sample t-test
t_statistic, p_value = stats.ttest_1samp(data, population_mean)
print("T-statistic:", t_statistic)
print("P-value:", p_value)
```

```python
import numpy as np
from scipy import stats
scores = [68,70,72,65,74,69,71,73,67,70]
# perform t-test
t_stat, p_value = stats.ttest_1samp(scores,70)
print("T-statistic:", t_stat)
print("P-value:", p_value)
```

Import Libraries

```python
import numpy as np
from scipy import stats
```

Define Data

```python
scores = [65,68,70,72,75,78,80,82,85,90]
mu = 75
```

Calculate Mean

```python
mean = np.mean(scores)
print("Mean:", mean)
```

Perform One-Sample t-Test

```python
t_stat, p_value = stats.ttest_1samp(scores, mu)

print("T-statistic:", t_stat)
print("P-value:", p_value)
```

Decision Rule

```python
alpha = 0.05

if p_value < alpha:
    print("Reject H0")
else:
    print("Do not reject H0")
```
