Data Cleaning & Preprocessing

fill missing values

---



```python
# Fill missing values if any
df.fillna(method='ffill', inplace=True)

# Convert Date column to datetime (if available)
if 'Date' in df.columns:
    df['Date'] = pd.to_datetime(df['Date'])
    df['Month'] = df['Date'].dt.month
```

```python
df.head()
```

EDA

Total and Average Expense

```python
df['Sales'] = pd.to_numeric(df['Sales'], errors='coerce')
df['Sales'].fillna(0, inplace=True)

total_expense = df['Sales'].sum()
average_expense = df['Sales'].mean()

print("Total Expense:", total_expense)
print("Average Expense:", average_expense)
```

EDA

Category-wise Expense Analysis

```python
category_expense = df.groupby('Product line')['Sales'].sum().sort_values(ascending=False)
category_expense
```

Detect Trends Over Time

```python
# Convert Date column to datetime
df['Date'] = pd.to_datetime(df['Date'])

# Group by Date
daily_expense = df.groupby('Date')['Sales'].sum()

# Line Plot
plt.figure(figsize=(10,5))
daily_expense.plot()
plt.title("Expense Trend Over Time")
plt.xlabel("Date")
plt.ylabel("Total Expense")
plt.show()
```

Identify Outliers in Spending

```python
import seaborn as sns

# Boxplot to detect outliers
plt.figure(figsize=(6,4))
sns.boxplot(y=df['Sales'])
plt.title("Outlier Detection in Spending")
plt.show()

# Optional: Show extreme values
Q1 = df['Sales'].quantile(0.25)
Q3 = df['Sales'].quantile(0.75)
IQR = Q3 - Q1

outliers = df[(df['Sales'] < (Q1 - 1.5 * IQR)) | (df['Sales'] > (Q3 + 1.5 * IQR))]

print("Outlier Records:")
print(outliers)
```
