2. Python Demonstration of Central Limit Theorem

```python
import numpy as np
import matplotlib.pyplot as plt
# create population data
population = np.random.uniform(0,100,10000)
sample_means = []
# take 1000 samples
for i in range(1000):
    sample = np.random.choice(population,30)
    sample_means.append(np.mean(sample))
# plot distribution of sample means
plt.hist(sample_means)
plt.title("Central Limit Theorem Demonstration")
plt.xlabel("Sample Means")
plt.ylabel("Frequency")
plt.show()
```

```python
# Homework: Sampling Distribution of the Mean

# Step 1: Generate 10,000 random numbers as population
import numpy as np
import matplotlib.pyplot as plt

# Let's assume population is normally distributed
population = np.random.randn(10000)  # mean=0, std=1

# Step 2: Take 500 samples (size = 30)
sample_size = 30
num_samples = 500

sample_means = []
for _ in range(num_samples):
    sample = np.random.choice(population, sample_size, replace=False)
    sample_means.append(np.mean(sample))

# Step 3: Calculate sample means
sample_means = np.array(sample_means)

# Step 4: Plot histogram of sample means
plt.figure(figsize=(8,6))
plt.hist(sample_means, bins=30, edgecolor='black', alpha=0.7)
plt.title("Histogram of Sample Means (n=30, 500 samples)")
plt.xlabel("Sample Mean")
plt.ylabel("Frequency")
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```

Explanation of Steps
- Population: We generated 10,000 random numbers from a normal distribution.
- Sampling: We drew 500 samples, each of size 30, from that population.
- Sample Means: For each sample, we calculated the mean.
- Histogram: We plotted the distribution of those sample means.
This demonstrates the Central Limit Theorem: even if the population distribution isn’t perfectly normal, the distribution of sample means tends to be approximately normal.
