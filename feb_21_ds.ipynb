```python
import pandas as pd

import numpy as np

np.random.seed(42)

data = {

    "Age": np.random.randint(21, 60, 100),

    "Salary": np.random.randint(30000, 120000, 100),

    "Experience": np.random.randint(1, 35, 100),

    "Department": np.random.choice(["HR", "IT", "Sales", "Finance"], 100),

    "Gender": np.random.choice(["Male", "Female"], 100),

    "PerformanceScore": np.random.randint(1, 11, 100)

}

df = pd.DataFrame(data)

df
```

```python
import matplotlib.pyplot as plt

import seaborn as sns
```

```python
# Histrogram plot

plt.figure(figsize=(6,4))

plt.hist(df["Age"], bins=10)

plt.title("Age Distribution")

plt.xlabel("Age")

plt.ylabel("Frequency")

plt.show()
```

```python
df.hist()
```

```python
# count plot

plt.figure(figsize=(6,4))

sns.countplot(x="Department", data=df)

plt.title("Employee Count by Department")

plt.show()
```

```python
# scatter plot

plt.figure(figsize=(6,4))

sns.scatterplot(x="Experience", y="Salary", data=df)

plt.title("Experience vs Salary")

plt.show()
```

```python
plt.figure(figsize=(6,4))

sns.scatterplot(x="Salary", y="PerformanceScore", data=df)

plt.title("Salary vs PerformanceScore")

plt.show()
```

```python
# boxplot ,we will check outline.

plt.figure(figsize=(6,4))

sns.boxplot(x="Department", y="Salary", data=df)

plt.title("Salary Distribution by Department")

plt.show()
```

```python
# pair plot

sns.pairplot(df)
plt.show
```

```python
# carrelation heat map

plt.figure(figsize=(7,5))

corr = df.corr(numeric_only=True)

sns.heatmap(corr, annot=True, cmap="coolwarm")

plt.title("Correlation Heatmap")

plt.show()
```
