```python
import numpy as np
import pandas as pd
```

```python
# Creating sample dataset
marks = [70, 80, 75, 90, 65, 85, 78, 88, 92, 74]
```

```python
# Convert into pandas series for analysis
marks_series = pd.Series(marks)
```

Mean (Average)

Formula:
\mu = \frac{\sum x}{n}

Where
• μ = Mean

• Σx = Sum of all values

• n = Number of observations

```python
70, 80, 75, 90, 65
Mean = (70 + 80 + 75 + 90 + 65) / 5
```

```python
# Calculate mean using numpy
mean_value = np.mean(marks)
```

```python
# Print mean
print("Mean:", mean_value)
```

Median

Steps:
1. Sort the data

2. Pick the middle value


```python
# Calculate median
median_value = np.median(marks)
print("Median:", median_value)
```

Mode

```python
from scipy import stats
# dataset example
data = [10, 20, 20, 30, 40]
# calculate mode
mode_value = stats.mode(data)
print("Mode:", mode_value)
```

Central Tendency Matters in ML

```python
# Create a sample DataFrame with missing values
data = {'Age': [20, 21, np.nan, 23, 24, np.nan, 26, 27, 28, 29],
        'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eve', 'Frank', 'Grace', 'Heidi', 'Ivan', 'Judy']}
df = pd.DataFrame(data)

# Replace missing values in the 'Age' column with the mean of the column
df["Age"].fillna(df["Age"].mean(), inplace=True)

print(df)
```

```python
import pandas as pd
data = {
"Salary": [30000, 35000, 40000, 42000, 30000, 38000, 36000]
}
df = pd.DataFrame(data)
print("Mean:", df["Salary"].mean())
print("Median:", df["Salary"].median())
print("Mode:", df["Salary"].mode())
```
