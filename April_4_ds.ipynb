Python Example (Simple Q-Learning Demonstration)

```python
# Import numpy library for numerical calculations
import numpy as np

# Number of states in our environment
states = 5

# Number of possible actions
actions = 2

# Initialize Q-table with zeros
Q = np.zeros((states, actions))

# Learning rate
alpha = 0.1

# Discount factor
gamma = 0.9

# Total training episodes
episodes = 1000

# Loop through each training episode
for episode in range(episodes):

    # Start from state 0
    state = 0

    # Continue until goal state is reached
    while state != 4:

        # Choose a random action (exploration)
        action = np.random.randint(actions)

        # Define next state based on action
        if action == 0:
            next_state = max(0, state - 1)
        else:
            next_state = min(states - 1, state + 1)

        # Reward system
        if next_state == 4:
            reward = 10
        else:
            reward = -1

        # Update Q-value using Q-learning formula
        Q[state, action] = Q[state, action] + alpha * (
            reward + gamma * np.max(Q[next_state]) - Q[state, action]
        )

        # Move to the next state
        state = next_state

# Print the final Q-table
print("Learned Q-table:")
print(Q)
```

Python Example: Detecting Bias in Dataset

```python
# Import pandas library for data manipulation
import pandas as pd
# Create a sample dataset
data = {
"Gender": ["Male","Male","Male","Female","Male","Male","Female","Male"]
}
# Convert dictionary to pandas DataFrame
df = pd.DataFrame(data)
# Count occurrences of each gender
gender_count = df["Gender"].value_counts()
# Print gender distribution
print(gender_count)
```

```python
# Import required libraries
from sklearn.ensemble import RandomForestClassifier
from sklearn.datasets import load_iris
# Load iris dataset
data = load_iris()
# Features
X = data.data
# Target labels
y = data.target
# Create Random Forest model
model = RandomForestClassifier()
# Train the model
model.fit(X, y)
# Get feature importance scores
importance = model.feature_importances_
# Print importance values
print("Feature Importance:", importance)
```
