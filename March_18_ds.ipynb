Range


```python
import numpy as np
marks = [60,65,70,75,80,85,90]
# calculate maximum value
max_value = np.max(marks)
# calculate minimum value
min_value = np.min(marks)
# calculate range
range_value = max_value - min_value
print("Range:", range_value)
```

Variance


```python
data = [10,12,14]
# calculate variance
variance_value = np.var(data)
print("Variance:", variance_value)
```

Standard Deviation

```python
data = [10,12,14]
# calculate standard deviation
std_value = np.std(data)
print("Standard Deviation:", std_value)
```

 Quartiles

```python
import numpy as np
data = [10,20,30,40,50,60,70,80]
# calculate quartiles
q1 = np.percentile(data,25)
q2 = np.percentile(data,50)
q3 = np.percentile(data,75)
print("Q1:", q1)
print("Median (Q2):", q2)
print("Q3:", q3)
```

Percentiles

```python
import numpy as np
data = [10,20,30,40,50,60,70,80,90]
# calculate 90th percentile
p90 = np.percentile(data,90)
print("90th Percentile:", p90)
```

Interquartile Range

```python
import numpy as np
data = [10,20,30,40,50,60,70,80]
q1 = np.percentile(data,25)
q3 = np.percentile(data,75)
iqr = q3 - q1
print("IQR:", iqr)
```

 Detecting Outliers Using IQR

```python
import numpy as np
data = [10,20,30,40,50,60,70,80,200]
q1 = np.percentile(data,25)
q3 = np.percentile(data,75)
iqr = q3 - q1
lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr
# identify outliers
outliers = [x for x in data if x < lower_bound or x > upper_bound]
print("Outliers:", outliers)
```

```python
# Dataset
data = [12,15,18,20,22,25,30,35,40,100]

import numpy as np
```

1. Range

```python
# 1. Range
range_val = max(data) - min(data)
print("Range:", range_val)
```

2. Variance

```python
# 2. Variance (Population variance)
variance = np.var(data)
print("Variance:", variance)
```

 3. Standard Deviation

```python
# 3. Standard Deviation (Population std)
std_dev = np.std(data)
print("Standard Deviation:", std_dev)
```

4. Quartiles

```python
# 4. Quartiles
Q1 = np.percentile(data, 25)
Median = np.percentile(data, 50)
Q3 = np.percentile(data, 75)
print("Q1:", Q1, "Median:", Median, "Q3:", Q3)
```

5. Interquartile Range

```python
# 5. Interquartile Range (IQR)
IQR = Q3 - Q1
print("IQR:", IQR)
```

6. Detect Outliers using Tukey’s Rule

```python
# 6. Detect Outliers using Tukey’s Rule
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
outliers = [x for x in data if x < lower_bound or x > upper_bound]
print("Outliers:", outliers)
```
